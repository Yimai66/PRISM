## PRISM Role 1 Technical Prototype
This notebook documents how to run the PRISM Streamlit prototype.

The main application is implemented in `app.py` because PRISM is an interactive web prototype rather than a traditional notebook-only workflow. The app allows users to compare search results across different language or national contexts and export the results for evaluation.

## Current Pipeline
The current stable version uses:

1. Case Book-based verified translations for the six demo questions  
2. Live Brave Search API for search results  
3. Live LLM disagreement detection using Role 2's six-type taxonomy  
4. CSV export for Role 4 evaluation  

The real `.env` file is not included because it contains API keys.

```python
print("Main app file: app.py")
print("To run the prototype locally:")
print("1. Install requirements: pip install -r requirements.txt")
print("2. Copy .env.example to .env and add API keys")
print("3. Run: streamlit run app.py")
print("4. Open: http://localhost:8501")
```
